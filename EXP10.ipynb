```python

# ============================================================
# EXPERIMENT: EMAIL AUTOMATION AGENT USING PYTHON
# Features:
# 1. Email Reading
# 2. Email Classification
# 3. Automated Reply Generation
# 4. Meeting Scheduling
# 5. Calendar Conflict Detection
# 6. Sequential Workflow Execution
# ============================================================

import re
from datetime import datetime, timedelta

# ============================================================
# 1. SAMPLE EMAIL INBOX
# ============================================================

emails = [
    {
        "id": 1,
        "sender": "rahul@gmail.com",
        "subject": "Project Discussion Meeting",
        "body": "Hi, can we schedule a meeting tomorrow at 10 AM to discuss the project?"
    },
    {
        "id": 2,
        "sender": "hr@company.com",
        "subject": "Important Internship Update",
        "body": "Please review the internship selection update. Your response is required."
    },
    {
        "id": 3,
        "sender": "student@gmail.com",
        "subject": "Question regarding assignment",
        "body": "Could you please explain the submission deadline for the assignment?"
    },
    {
        "id": 4,
        "sender": "unknown123@gmail.com",
        "subject": "Congratulations! You won a prize",
        "body": "Click this link immediately to claim your free prize and reward!"
    }
]

# ============================================================
# 2. SIMULATED CALENDAR
# ============================================================

calendar = [
    {
        "date": "2026-11-06",
        "start": "10:00",
        "end": "11:00",
        "title": "Team Meeting"
    }
]

# ============================================================
# 3. EMAIL READER AGENT
# ============================================================

class EmailReaderAgent:

    def read_emails(self, inbox):
        print("\n📩 EMAIL READER AGENT")
        print("-" * 50)

        for email in inbox:
            print(f"Email ID: {email['id']}")
            print(f"From: {email['sender']}")
            print(f"Subject: {email['subject']}")
            print(f"Body: {email['body']}")
            print("-" * 50)

        return inbox


# ============================================================
# 4. EMAIL CLASSIFIER AGENT
# ============================================================

class EmailClassifierAgent:

    def classify(self, email):

        text = (email["subject"] + " " + email["body"]).lower()

        spam_keywords = [
            "won a prize", "free prize", "claim your reward",
            "click this link", "lottery", "congratulations"
        ]

        meeting_keywords = [
            "meeting", "schedule", "discuss", "tomorrow",
            "appointment", "call", "conference"
        ]

        important_keywords = [
            "urgent", "important", "required",
            "deadline", "action required", "update"
        ]

        if any(word in text for word in spam_keywords):
            return "Spam"

        elif any(word in text for word in meeting_keywords):
            return "Meeting"

        elif any(word in text for word in important_keywords):
            return "Important"

        else:
            return "Query"


# ============================================================
# 5. AUTOMATED REPLY AGENT
# ============================================================

class ReplyGeneratorAgent:

    def generate_reply(self, email, category):

        if category == "Meeting":
            reply = (
                f"Hello,\n\nThank you for your email regarding "
                f"'{email['subject']}'. I will check my availability "
                "and confirm the meeting.\n\nRegards"
            )

        elif category == "Important":
            reply = (
                f"Hello,\n\nI have received your important email "
                f"regarding '{email['subject']}'. "
                f"I will review it and respond shortly.\n\nRegards"
            )

        elif category == "Query":
            reply = (
                f"Hello,\n\nThank you for your query regarding "
                f"'{email['subject']}'. I will review your question "
                f"and get back to you soon.\n\nRegards"
            )

        else:
            reply = "No reply generated. This email was classified as spam."

        return reply


# ============================================================
# 6. MEETING SCHEDULER AGENT
# ============================================================

class MeetingSchedulerAgent:

    def find_meeting_time(self, email, existing_calendar):

        text = email["body"].lower()

        # Detect meeting request
        if not any(word in text for word in
                   ["meeting", "schedule", "discuss", "appointment"]):
            return None

        # Extract date and time from sample email
        if "tomorrow" in text:
            meeting_date = datetime.now() + timedelta(days=1)
        else:
            meeting_date = datetime.now()

        # Extract hour (e.g., 10 AM, 2 PM)
        match = re.search(
            r"(\d{1,2})\s*(am|pm)",
            text
        )

        if match:
            hour = int(match.group(1))
            period = match.group(2)

            if period == "pm" and hour != 12:
                hour += 12
            elif period == "am" and hour == 12:
                hour = 0
        else:
            hour = 14  # Default time: 2 PM

        start_time = meeting_date.replace(
            hour=hour, minute=0, second=0, microsecond=0
        )

        end_time = start_time + timedelta(hours=1)

        # Check for calendar conflicts
        for event in existing_calendar:

            event_date = datetime.strptime(
                event["date"], "%Y-%m-%d"
            )

            event_start = datetime.strptime(
                event["start"], "%H:%M"
            )

            event_end = datetime.strptime(
                event["end"], "%H:%M"
            )

            existing_start = event_date.replace(
                hour=event_start.hour,
                minute=event_start.minute
            )

            existing_end = event_date.replace(
                hour=event_end.hour,
                minute=event_end.minute
            )

            # Overlapping event check
            if (
                start_time.date() == existing_start.date()
                and start_time < existing_end
                and end_time > existing_start
            ):
                return {
                    "status": "Conflict",
                    "message": "Requested time conflicts with an existing event."
                }

        return {
            "status": "Available",
            "date": start_time.strftime("%Y-%m-%d"),
            "start": start_time.strftime("%H:%M"),
            "end": end_time.strftime("%H:%M"),
            "title": email["subject"]
        }


# ============================================================
# 7. TASK EXECUTION AGENT
# ============================================================

class TaskExecutionAgent:

    def execute(self, email, category, reply, meeting):

        print("\n⚙️ TASK EXECUTION AGENT")
        print("-" * 50)

        print("Email Category:", category)

        if category == "Spam":
            print("Action: Email marked as spam.")
            print("Reply: Not generated.")
            return

        print("\nGenerated Automated Reply:")
        print(reply)

        if category == "Meeting":

            if meeting is None:
                print("\nNo meeting details detected.")
                return

            if meeting["status"] == "Conflict":
                print("\n⚠️ Meeting Scheduling Failed")
                print(meeting["message"])

            else:
                print("\n📅 Meeting Scheduled Successfully!")
                print("Title:", meeting["title"])
                print("Date:", meeting["date"])
                print("Start:", meeting["start"])
                print("End:", meeting["end"])

                # Add event to simulated calendar
                calendar.append(meeting)

        print("-" * 50)


# ============================================================
# 8. MAIN EMAIL AUTOMATION WORKFLOW
# ============================================================

class EmailAutomationAgent:

    def __init__(self):

        self.reader = EmailReaderAgent()
        self.classifier = EmailClassifierAgent()
        self.reply_agent = ReplyGeneratorAgent()
        self.scheduler = MeetingSchedulerAgent()
        self.executor = TaskExecutionAgent()

    def run(self, inbox):

        print("\n🤖 EMAIL AUTOMATION AGENT STARTED")
        print("=" * 60)

        # Step 1: Read emails
        received_emails = self.reader.read_emails(inbox)

        # Step 2: Process emails sequentially
        for email in received_emails:

            print("\n\n🔄 PROCESSING EMAIL ID:", email["id"])

            # Step 3: Classify email
            category = self.classifier.classify(email)
            print("📌 Classification:", category)

            # Step 4: Generate reply
            reply = self.reply_agent.generate_reply(
                email, category
            )

            # Step 5: Schedule meeting if needed
            meeting = None

            if category == "Meeting":
                meeting = self.scheduler.find_meeting_time(
                    email, calendar
                )

            # Step 6: Execute appropriate actions
            self.executor.execute(
                email, category, reply, meeting
            )

        print("\n✅ EMAIL AUTOMATION WORKFLOW COMPLETED")
        print("=" * 60)


# ============================================================
# 9. RUN THE AGENT
# ============================================================

agent = EmailAutomationAgent()
agent.run(emails)
```
